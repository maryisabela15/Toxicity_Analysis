```python
import pandas as pd
import re

data = pd.read_csv("train.csv")

data.head()
```

```python
# Create a function to clean the data in the column "comment_text"

def clean_text(text):
    text = str(text)
    text = re.sub(r'\s+', ' ', text)  # Normalize whitespace
    text = text.strip()               # Remove leading/trailing spaces
    text = re.sub(r'<.*?>', '', text) # Remove HTML tags
    return text
```

```python
# Apply cleaning
data['text'] = data['comment_text'].apply(clean_text)

# Keep only the 'text' column for annotation
clean_data = data[['text']]

# (Optional) Remove very short comments
clean_data = clean_data[clean_data['text'].str.split().str.len() >= 3]

# Save the cleaned file
clean_data.to_csv('cleaned_comments.csv', index=False)

print("✅ Cleaned dataset saved as 'cleaned_comments.csv'")
```
